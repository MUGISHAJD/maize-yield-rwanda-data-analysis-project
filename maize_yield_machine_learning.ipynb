```python
#load libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#load the dataset
df = pd.read_csv("maize_yield_rwanda_rw.csv")
#df.head(20)
```

```python
#Display number of rows and column before data cleaning
print("NUMBER OF ROWS AND COLUMNS")
rows, cols = df.shape
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])    
```

```python
#Check missing value by indicatind number od null values and their percentages. 
missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})
missing = missing.sort_values(
    by="Missing_Percentage",
    ascending=False
)

#display(missing)
```

```python
#Visualizing the missing values 
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap=['#f0f0f0', '#d9534f']
)

plt.title("Missing Values in Maize Dataset")
plt.xlabel("Columns")
plt.show()
```

```python
# Check complete duplicate rows
full_duplicates = df.duplicated().sum()

# Check repeated farm IDs
duplicate_farms = df["farm_id"].duplicated().sum()

print("Complete duplicate rows:", full_duplicates)
print("Repeated farm IDs:", duplicate_farms)
```
